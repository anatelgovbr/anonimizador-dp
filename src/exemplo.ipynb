# Importe a bilioteca

````python
try:
    from anonimizar.sei_anonimizar import SeiAnonimizar
except:
    msg = "Você deve instalar o anonimizador para poder usa-lo. Siga as instruções da documentação"
    raise ModuleNotFoundError(msg)
````

# Defina o caminho do modelo

````python
model_path = "Seu_caminho"
````

Voce ainda pode definir um caminho na variavel de ambiente 
```python
from dotenv import load_dotenv

load_dotenv()

model_path = os.getenv("SPACY_MODEL_PATH")
```

# Texto exemplo

````python
texto_original = """
Relatório do cliente:
Nome: João Silva
CPF: 123.456.789-00
RG: 12.345.678-9
Telefone: (61) 9999-8888
"""
````

# Instancie o anonimizador

````python
anon = SeiAnonimizar(model_path=model_path)
anon.add_apply_patterns(["CPF", "RG", "TELEFONE"])
````

## Busque as entidades

````python
doc_ents = anon.extract_entities(text_or_path=texto_original, return_type="label_position")
print(doc_ents)
````

## Adicionando seu proprio padrao regex
pode realizar os testes inicialmente no [Regex_101](https://regex101.com/)

````python
anon.add_apply_patterns(
    labels=["CPF", "EMAIL"],
    custom_patterns=[
        {
            "label": "MATRICULA_ALUNO",
            "regex": r"[A-Z]{2}\d{5}[A-Z]",
            "description": "Matrícula de aluno (2 letras + 5 números + 1 letra)"
        },
        {
            "label": "PLACA_CARRO",
            "regex": r"[A-Z]{3}-\d{4}",
            "description": "Placa de carro no formato AAA-9999"
        }
    ]
)
````

````python
texto_original = "A placa no veiculo é ABC-1234"
````

````python
doc_ents = anon.extract_entities(text_or_path=texto_original, return_type="label_position")
print(doc_ents)
````

## Anonimize textos

````python
texto_original = """
--- Dados do cadastro ---
Nome: Zezinho Sauro
CPF: 123.456.789-12
Telefone: (61) 9999-8888
E-mail: ze_o_dinossauro@hotmail.com
"""
````

````python
doc_ents = anon.extract_entities(text_or_path=texto_original, return_type="label_position")
texto_anonimizado = anon.anonymize_text(texto_original, doc_ents)
print(texto_anonimizado)
````
