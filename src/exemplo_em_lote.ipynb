# Importe a bilioteca

````python
try:
    from anonimizar.sei_anonimizar import SeiAnonimizar
except:
    msg = "Você deve instalar o anonimizador para poder usa-lo. Siga as instruções da documentação"
    raise ModuleNotFoundError(msg)
````

# Defina o caminho do modelo

````python
model_path = "Seu_caminho"
````

Voce ainda pode definir um caminho na variavel de ambiente 
```python
from dotenv import load_dotenv

load_dotenv()

model_path = os.getenv("SPACY_MODEL_PATH")
```

# Lista de textos para exemplo

````python
lista_textos = [
    """
Relatório do cliente:
Nome: João Silva
CPF: 123.456.789-00
RG: 12.345.678-9
Telefone: (61) 9999-8888
    """,
    """
Ola, meu nome é C3PO e meu CPF é 000.000.000-00
    """,
]
````

# Instancie o anonimizador

````python
anon = SeiAnonimizar(model_path=model_path)
anon.add_apply_patterns(["CPF", "RG", "TITULO_ELEITOR",
"PASSAPORTE", "SIAPE", "TELEFONE", "DATA_NASCIMENTO", "ENDEREÇO",
"DADOS_BANCARIOS", "EMAIL", "CNH"])
````

## Busque as entidades

````python
entitities = []
for texto_original in lista_textos:
    doc_ents = anon.extract_entities(text_or_path=texto_original, return_type="label_position")
    entitities.append(doc_ents)
    print(doc_ents)
````
