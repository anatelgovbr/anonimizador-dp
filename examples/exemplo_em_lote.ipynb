# Defina o caminho do modelo

````python
# ruff: noqa: T201
import os

model_path = os.getenv("SPACY_MODEL_PATH")
if not model_path:
    error_message = "Defina SPACY_MODEL_PATH com o caminho de um modelo spaCy."
    raise RuntimeError(error_message)
````

# Lista de textos para exemplo

````python
lista_textos = [
    """
RelatÃ³rio do cliente:
Nome: JoÃ£o Silva
CPF: 123.456.789-09
RG: 12.345.678-9
Telefone: (61) 9999-8888
    """,
    """
Ola, meu nome Ã© C3PO e meu CPF Ã© 123.456.789-09
    """,
]
````

# Instancie o anonimizador

````python
anon = Anonimizar(model_path=model_path, auto_patterns=False)
anon.add_apply_patterns(
    [
        "CPF",
        "RG",
        "TITULO_ELEITOR",
        "PASSAPORTE",
        "SIAPE",
        "TELEFONE",
        "DATA_NASCIMENTO",
        "ENDEREÃ‡O",
        "DADOS_BANCARIOS",
        "EMAIL",
        "CNH",
    ]
)
````

## Busque as entidades

````python
# ruff: noqa: T201
resultados = []
for indice, texto_original in enumerate(lista_textos):
    doc_ents = anon.extract_entities(text_or_path=texto_original, return_type="label_position")
    texto_anonimizado = anon.anonymize_text(texto_original, doc_ents)
    resultados.append({"id": indice, "texto": texto_original, "entidades": doc_ents, "anonimizado": texto_anonimizado})
    print(resultados[-1])
````
