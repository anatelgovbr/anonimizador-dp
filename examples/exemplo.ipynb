# Defina o caminho do modelo

````python
# ruff: noqa: T201
import os

model_path = os.getenv("SPACY_MODEL_PATH")
if not model_path:
    error_message = "Defina SPACY_MODEL_PATH com o caminho de um modelo spaCy."
    raise RuntimeError(error_message)
````

Voce ainda pode definir um caminho na variavel de ambiente 
```python

import os

model_path = os.getenv("SPACY_MODEL_PATH")

    raise RuntimeError("Defina SPACY_MODEL_PATH com o caminho de um modelo spaCy.")
```

# Texto exemplo

````python
texto_original = """
RelatÃ³rio do cliente:
Nome: JoÃ£o Silva
CPF: 123.456.789-09
RG: 12.345.678-9
Telefone: (61) 9999-8888
"""
````

# Instancie o anonimizador

````python
anon = Anonimizar(model_path=model_path, auto_patterns=False)
anon.add_apply_patterns(["CPF", "RG", "TELEFONE"])
````

## Busque as entidades

````python
# ruff: noqa: T201
doc_ents = anon.extract_entities(text_or_path=texto_original, return_type="label_position")
print(doc_ents)
````

## Anonimize textos

````python
texto_original = """
--- Dados do cadastro ---
Nome: Zezinho Sauro
CPF: 123.456.789-09
Telefone: (61) 9999-8888
E-mail: ze_o_dinossauro@hotmail.com
"""
````

````python
# ruff: noqa: T201
doc_ents = anon.extract_entities(text_or_path=texto_original, return_type="label_position")
texto_anonimizado = anon.anonymize_text(texto_original, doc_ents)
print(texto_anonimizado)
````
